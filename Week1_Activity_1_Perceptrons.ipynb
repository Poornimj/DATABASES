# Activity 1: Perceptrons, Logic Gates, and the XOR Limitation

## 1. Implement the Perceptron

### Predictions before coding

Inputs: [1, 0]
Weights: [0.5, 0.5]

With bias -0.7:
z = (1 × 0.5) + (0 × 0.5) - 0.7 = -0.2
The output is 0 because z is negative.

With bias -0.3:
z = (1 × 0.5) + (0 × 0.5) - 0.3 = 0.2
The output is 1 because z is positive.

```python
import numpy as np
import matplotlib.pyplot as plt


def perceptron(inputs, weights, bias):
    """Return 0 or 1 using a step activation."""
    assert len(inputs) == len(weights), "Inputs and weights must have equal length."

    weighted_sum = np.dot(inputs, weights) + bias

    if weighted_sum >= 0:
        return 1
    else:
        return 0
```

```python
print("First output:", perceptron([1, 0], [0.5, 0.5], -0.7))
print("Changed bias:", perceptron([1, 0], [0.5, 0.5], -0.3))

assert perceptron([1, 0], [0.5, 0.5], -0.7) == 0
assert perceptron([1, 0], [0.5, 0.5], -0.3) == 1

print("Both tests passed!")
```

### Explanation: Effect of changing the bias

The inputs and weights stayed the same, so their weighted sum was 0.5.
Changing the bias from -0.7 to -0.3 changed the final value from -0.2
to 0.2. The step activation returns 0 for a negative value and 1 for
a value greater than or equal to zero. Therefore, changing only the
bias changed the output.

## 2. Model Logic Gates
### Task 2A: OR gate

OR outputs 1 when at least one input is 1.

I chose weights [1.0, 1.0] and bias -0.5.
The calculation is z = x1 + x2 - 0.5.

My predicted outputs for [0,0], [0,1], [1,0], and [1,1]
are [0, 1, 1, 1].

```python
gate_inputs = [[0, 0], [0, 1], [1, 0], [1, 1]]
expected_or = [0, 1, 1, 1]

weights_or = [1.0, 1.0]
bias_or = -0.5

actual_or = [perceptron(pair, weights_or, bias_or) for pair in gate_inputs]

for pair, actual, expected in zip(gate_inputs, actual_or, expected_or):
    print(f"Input: {pair} | Predicted: {actual} | Expected: {expected}")

assert actual_or == expected_or, "Your OR gate does not yet match its truth table."
print("OR gate works!")
```
